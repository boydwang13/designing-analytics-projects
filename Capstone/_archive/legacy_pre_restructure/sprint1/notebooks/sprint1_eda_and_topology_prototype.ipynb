```python
from pathlib import Path
import pandas as pd

base = Path('..').resolve() if Path.cwd().name == 'notebooks' else Path('Capstone')
nodes = pd.read_csv(base / 'mock_data_nodes.csv')
edges = pd.read_csv(base / 'mock_data_edges.csv')
print({'nodes': len(nodes), 'edges': len(edges)})
```

```python
display(edges['Interaction_Frequency'].value_counts())
display(edges['Interaction_Type'].value_counts())
```

```python
in_deg = edges.groupby('Target_EMP_ID').size().rename('in_degree')
out_deg = edges.groupby('Source_EMP_ID').size().rename('out_degree')
display(in_deg.sort_values(ascending=False).head(10))
display(out_deg.sort_values().head(10))
```

```python
# DQ checks (Sprint 1 quick gate)
required_cols = {
    'Source_EMP_ID','Target_EMP_ID','Interaction_Type','Interaction_Frequency',
    'Awareness_Score','Energy_Score','Nomination_Rank'
}

assert set(edges.columns) >= required_cols, 'Missing required edge columns'
assert edges['Interaction_Type'].isin(['Hard', 'Soft']).all(), 'Invalid Interaction_Type found'
assert edges['Interaction_Frequency'].isin(['Daily', 'Weekly', 'Monthly', 'Rarely']).all(), 'Invalid frequency found'
assert edges['Awareness_Score'].between(1, 5).all(), 'Awareness_Score out of range'
assert edges['Energy_Score'].between(1, 5).all(), 'Energy_Score out of range'
assert edges['Nomination_Rank'].between(1, 3).all(), 'Nomination_Rank out of range'

dup_rank = edges.duplicated(['Source_EMP_ID', 'Interaction_Type', 'Nomination_Rank']).sum()
print({'dq_pass': dup_rank == 0, 'duplicate_rank_count': int(dup_rank)})
```
